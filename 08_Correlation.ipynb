# 08_Correlation

### What is Correlation?

Correlation measures the strength and direction of the linear relationship between two variables. It tells us how much and in which direction one variable tends to change when the other variable changes.

Example

If ice cream sales and temperature both rise together, they are positively correlated. If umbrella sales rise when sunshine hours fall, they are negatively correlated.

## 01. Correlation

**Definition**: Correlation quantifies the degree to which two variables move together, on a scale from -1 to +1, where the sign indicates direction and the magnitude indicates strength.

**Formula**: r = Cov(X,Y) / (σx × σy)

**Where it is used**: Used to identify relationships between variables before building models, in feature selection, and in exploratory data analysis.

**Real-World Example**: Height and weight of people tend to be positively correlated.

**AI/ML Example**: Correlation is used to detect multicollinearity between input features before training a regression model.

```python
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
corr = np.corrcoef(x, y)[0,1]
print(corr)
```

**Interpretation**: The correlation value of 1.0 shows x and y increase together in a perfect straight-line relationship.

## 02. Covariance

**Definition**: Covariance measures the direction of the linear relationship between two variables by showing whether they increase or decrease together, but unlike correlation, its magnitude is not standardized.

**Formula**: Cov(X,Y) = Σ[(xi − x̄)(yi − ȳ)] / (n − 1)

**Where it is used**: Used as the building block for correlation, and in portfolio variance calculations in finance.

**Real-World Example**: Covariance between stock returns of two companies shows whether they move in the same or opposite direction.

**AI/ML Example**: Covariance matrices are used in Principal Component Analysis (PCA) to identify directions of maximum variance in the data.

```python
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
cov = np.cov(x, y)[0,1]
print(cov)
```

**Interpretation**: The positive covariance confirms x and y move in the same direction, but its unscaled value makes it hard to judge relationship strength directly.

## 03. Positive Correlation

**Definition**: Positive correlation occurs when two variables move in the same direction, meaning as one increases, the other also increases.

**Formula**: r > 0, where r is the Pearson correlation coefficient

**Where it is used**: Used to identify variables that reinforce each other, useful in feature engineering and trend analysis.

**Real-World Example**: Study hours and exam scores usually show positive correlation.

**AI/ML Example**: In recommendation systems, positive correlation between user ratings of similar products helps identify items to recommend together.

```python
import numpy as np
hours = np.array([1,2,3,4,5,6])
scores = np.array([50,55,65,70,78,85])
corr = np.corrcoef(hours, scores)[0,1]
print(corr)
```

**Interpretation**: The strong positive correlation confirms that more study hours are associated with higher exam scores.

## 04. Negative Correlation

**Definition**: Negative correlation occurs when two variables move in opposite directions, meaning as one increases, the other decreases.

**Formula**: r < 0, where r is the Pearson correlation coefficient

**Where it is used**: Used to detect inverse relationships, useful in risk analysis and diversification strategies.

**Real-World Example**: The number of hours spent watching TV and exam scores often show negative correlation.

**AI/ML Example**: In churn prediction, negative correlation between customer engagement score and churn probability helps identify at-risk customers.

```python
import numpy as np
tv_hours = np.array([1,2,3,4,5,6])
scores = np.array([90,82,75,68,60,52])
corr = np.corrcoef(tv_hours, scores)[0,1]
print(corr)
```

**Interpretation**: The strong negative correlation confirms that more TV hours are associated with lower exam scores.

## 05. No Correlation

**Definition**: No correlation (or zero correlation) means there is no consistent linear relationship between two variables; changes in one variable do not predict changes in the other.

**Formula**: r ≈ 0, where r is the Pearson correlation coefficient

**Where it is used**: Used to rule out irrelevant features in a dataset before modeling.

**Real-World Example**: Shoe size and intelligence score typically show no correlation.

**AI/ML Example**: Features showing no correlation with the target variable are often dropped during feature selection to reduce noise.

```python
import numpy as np
np.random.seed(42)
shoe_size = np.random.randint(6,12,50)
iq_score = np.random.randint(90,130,50)
corr = np.corrcoef(shoe_size, iq_score)[0,1]
print(corr)
```

**Interpretation**: The correlation value near zero confirms shoe size and IQ score have no meaningful linear relationship.

## 06. Pearson Correlation

**Definition**: Pearson Correlation measures the strength and direction of the linear relationship between two continuous variables, assuming the data is normally distributed and the relationship is linear.

**Formula**: r = Σ[(xi − x̄)(yi − ȳ)] / √[Σ(xi − x̄)² × Σ(yi − ȳ)²]

**Where it is used**: Used when both variables are continuous and the relationship is expected to be linear.

**Real-World Example**: Pearson correlation between advertising spend and sales revenue in a business.

**AI/ML Example**: Pearson correlation is used to check linear relationships between numerical features and the target variable in a regression problem.

```python
from scipy.stats import pearsonr
x = [10,20,30,40,50]
y = [15,18,25,31,40]
corr, p_value = pearsonr(x, y)
print(corr)
print(p_value)
```

**Interpretation**: The high Pearson correlation coefficient with a low p-value confirms a strong and statistically significant linear relationship.

## 07. Spearman Correlation

**Definition**: Spearman Correlation measures the strength and direction of the monotonic relationship between two variables by using the ranks of the data rather than raw values, making it suitable for non-linear monotonic relationships or ordinal data.

**Formula**: rs = 1 − (6 × Σdi²) / [n(n² − 1)], where di = difference between ranks of xi and yi

**Where it is used**: Used when data is not normally distributed, contains outliers, or the relationship is monotonic but not linear.

**Real-World Example**: Spearman correlation between customer satisfaction ranking and product ranking.

**AI/ML Example**: Spearman correlation is used to evaluate ranking-based recommendation systems where relative order matters more than exact values.

```python
from scipy.stats import spearmanr
x = [1,2,3,4,5]
y = [2,1,4,3,5]
corr, p_value = spearmanr(x, y)
print(corr)
print(p_value)
```

**Interpretation**: The moderate Spearman correlation indicates a fairly consistent, though not perfect, monotonic relationship between the rankings of x and y.

## 08. Correlation Matrix

**Definition**: A Correlation Matrix is a table showing the pairwise correlation coefficients between multiple variables in a dataset, allowing quick identification of relationships across all variable pairs at once.

**Formula**: Matrix M where M[i][j] = correlation(Xi, Xj)

**Where it is used**: Used in exploratory data analysis to detect multicollinearity and understand relationships across many features simultaneously.

**Real-World Example**: A correlation matrix of a house price dataset shows how area, number of rooms, and location score relate to each other and to price.

**AI/ML Example**: Before training a regression model, data scientists examine the correlation matrix to drop highly correlated redundant features and avoid multicollinearity.

```python
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
    'area': np.random.randint(500,3500,100),
    'rooms': np.random.randint(1,6,100),
    'age': np.random.randint(0,30,100)
})
df['price'] = df['area']*300 + df['rooms']*50000 - df['age']*1000 + np.random.normal(0,10000,100)
corr_matrix = df.corr()
print(corr_matrix)
```

**Interpretation**: The matrix shows price is strongly positively correlated with area, moderately correlated with rooms, and negatively correlated with age.

## 09. Heatmap

**Definition**: A Heatmap is a graphical representation of a matrix (such as a correlation matrix) where individual values are represented by colors, making it easy to visually spot strong positive, negative, or weak relationships across many variables.

**Formula**: No separate formula; it visualizes an existing matrix using a color scale.

**Where it is used**: Used to visually communicate correlation matrices, confusion matrices, and other tabular numeric data.

**Real-World Example**: A heatmap of a hospital dataset instantly shows which patient vitals are most correlated with disease risk.

**AI/ML Example**: Data scientists use correlation heatmaps to quickly decide which features to drop due to high multicollinearity before feeding data into a machine learning model.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
df = pd.DataFrame({
    'area': np.random.randint(500,3500,100),
    'rooms': np.random.randint(1,6,100),
    'age': np.random.randint(0,30,100)
})
df['price'] = df['area']*300 + df['rooms']*50000 - df['age']*1000 + np.random.normal(0,10000,100)
corr_matrix = df.corr()
plt.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(corr_matrix)), corr_matrix.columns, rotation=45)
plt.yticks(range(len(corr_matrix)), corr_matrix.columns)
for i in range(len(corr_matrix)):
    for j in range(len(corr_matrix)):
        plt.text(j, i, round(corr_matrix.iloc[i,j],2), ha='center', va='center')
plt.title('Correlation Heatmap')
plt.show()
```

**Interpretation**: The heatmap visually confirms price has strong warm-colored correlation with area, moderate correlation with rooms, and cool-colored negative correlation with age.

## Overall Advantages and Limitations of Correlation & Covariance

### Advantages
1. Quickly identifies relationships between variables without building a full model.
2. Correlation is standardized (-1 to 1), making it easy to compare across different variable pairs.
3. Helps in feature selection by identifying redundant or irrelevant features.
4. Heatmaps and correlation matrices provide fast visual summaries for large datasets.
5. Forms the mathematical foundation for techniques like PCA and regression diagnostics.

### Limitations
1. Correlation only measures linear (Pearson) or monotonic (Spearman) relationships, missing complex non-linear patterns.
2. Correlation does not imply causation; two variables can be correlated without one causing the other.
3. Covariance values are unbounded and hard to interpret without standardization.
4. Sensitive to outliers, which can distort the correlation coefficient significantly.
5. High correlation between features (multicollinearity) can be a warning sign but doesn't say which feature to remove.

## Interview Questions & Answers

**1. What is the difference between correlation and covariance?**
Covariance shows the direction of the relationship between two variables but is unscaled and hard to interpret, while correlation standardizes this value between -1 and 1, making the strength of the relationship easy to compare across different variable pairs.

**2. Does correlation imply causation? Explain with an example.**
No, correlation does not imply causation; for example, ice cream sales and drowning incidents are positively correlated because both increase in summer, but ice cream does not cause drowning, a hidden factor (temperature) drives both.

**3. When would you use Spearman correlation instead of Pearson correlation?**
Spearman correlation is used when the data is not normally distributed, contains outliers, or the relationship between variables is monotonic but not strictly linear, since it works on ranks rather than raw values.

**4. What does a correlation coefficient close to 0 mean?**
A correlation coefficient close to 0 means there is little to no linear relationship between the two variables, though a non-linear relationship could still exist.

**5. Why is a correlation heatmap useful in the feature selection process of machine learning?**
A correlation heatmap allows data scientists to instantly visualize relationships across all feature pairs at once, making it easy to spot highly correlated (redundant) features that can be dropped to reduce multicollinearity before model training.

**6. What is multicollinearity and how is correlation used to detect it?**
Multicollinearity occurs when two or more input features are highly correlated with each other, which can destabilize regression coefficients, and it is detected by examining the correlation matrix for coefficients close to +1 or -1 between predictor variables.

**7. Can covariance be negative, and what does that mean?**
Yes, covariance can be negative, and a negative value means the two variables tend to move in opposite directions, meaning as one increases, the other tends to decrease.
