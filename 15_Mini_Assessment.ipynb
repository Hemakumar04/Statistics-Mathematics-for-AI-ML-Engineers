## Part A: Conceptual Questions

**Q1.** Explain the difference between Mean and Median.

**Answer:** The mean is the arithmetic average of all values in a dataset (sum of values divided by count), while the median is the middle value when the data is sorted in order. The mean is sensitive to outliers and skewed data, whereas the median is robust to extreme values, making it a better measure of central tendency for skewed distributions.

**Q2.** What is Variance?

**Answer:** Variance measures how spread out the values in a dataset are from the mean, calculated as the average of the squared differences between each value and the mean. A higher variance means the data points are more spread out; a lower variance means they are clustered closely around the mean.

**Formula:** Var(X) = Σ(xi − x̄)² / n (population) or / (n − 1) (sample)

**Q3.** What is Standard Deviation?

**Answer:** Standard deviation is the square root of variance, expressing the spread of data in the same units as the original data (rather than squared units), which makes it easier to interpret directly.

**Formula:** σ = √Var(X)

**Q4.** Explain Bayes' Theorem with an example.

**Answer:** Bayes' Theorem describes how to update the probability of an event based on new evidence. It relates the conditional probability P(A|B) to P(B|A), P(A), and P(B).

**Formula:** P(A|B) = [P(B|A) × P(A)] / P(B)

**Example:** Suppose 1% of a population has a disease (P(Disease) = 0.01). A test correctly detects the disease 99% of the time (P(Positive|Disease) = 0.99) but also gives a false positive 5% of the time for healthy people (P(Positive|No Disease) = 0.05). If someone tests positive, Bayes' Theorem lets us calculate the actual probability they have the disease, which turns out to be much lower than 99% because the disease is rare — this is a classic example showing why base rates matter.

**Q5.** Difference between Population and Sample.

**Answer:** A population is the complete set of all individuals or items relevant to a study, while a sample is a smaller subset drawn from that population, used to make inferences about the whole population when studying every member isn't practical. Statistics calculated from a sample (like sample mean or sample variance) are estimates of the true population parameters.

**Q6.** Explain Correlation and Covariance.

**Answer:** Covariance measures the direction of the linear relationship between two variables (positive, negative, or none) but is unscaled, making its magnitude hard to interpret. Correlation standardizes covariance by dividing it by the product of the two variables' standard deviations, producing a value between -1 and +1 that indicates both the direction and strength of the linear relationship, and is easy to compare across different variable pairs.

**Formula:** r = Cov(X,Y) / (σx × σy)

**Q7.** What is Skewness?

**Answer:** Skewness measures the asymmetry of a data distribution around its mean. Positive (right) skew means the tail extends toward higher values (mean > median), negative (left) skew means the tail extends toward lower values (mean < median), and zero skew indicates a symmetric distribution like the normal distribution.

**Q8.** Explain IQR.

**Answer:** The Interquartile Range (IQR) is a measure of statistical spread equal to the difference between the 75th percentile (Q3) and the 25th percentile (Q1), representing the range of the middle 50% of the data. It is commonly used to detect outliers, since values falling below Q1 − 1.5×IQR or above Q3 + 1.5×IQR are typically flagged as outliers.

**Formula:** IQR = Q3 − Q1

**Q9.** What is a p-value?

**Answer:** A p-value is the probability of observing a result at least as extreme as the one obtained, assuming the null hypothesis is true. A small p-value (typically < 0.05) suggests the observed data would be unlikely under the null hypothesis, leading researchers to reject it in favor of the alternative hypothesis; it does not measure the probability that the null hypothesis itself is true.

**Q10.** Difference between Type I and Type II Error.

**Answer:** A Type I error occurs when a true null hypothesis is incorrectly rejected (a "false positive"), with its probability denoted by α (the significance level). A Type II error occurs when a false null hypothesis is incorrectly accepted/not rejected (a "false negative"), with its probability denoted by β. There is typically a trade-off between the two: reducing one tends to increase the other for a fixed sample size.

## Part B: Coding Exercises

**C1.** Write Python code to calculate Variance for the array `data = [12, 15, 14, 10, 18, 20, 16, 13]`.

```python
import numpy as np

data = np.array([12, 15, 14, 10, 18, 20, 16, 13])

variance_population = np.var(data)          # population variance (ddof=0)
variance_sample = np.var(data, ddof=1)       # sample variance (ddof=1)

print("Population Variance:", variance_population)
print("Sample Variance:", variance_sample)
```

**C2.** Detect outliers in the array `data = [12, 15, 14, 10, 18, 20, 16, 13, 95, 11]` using the IQR method.

```python
import numpy as np

data = np.array([12, 15, 14, 10, 18, 20, 16, 13, 95, 11])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

**C3.** Calculate the correlation between the two variables `x = [1, 2, 3, 4, 5]` and `y = [5, 4, 3, 2, 1]`.

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 3, 2, 1])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation between x and y:", correlation)
```
