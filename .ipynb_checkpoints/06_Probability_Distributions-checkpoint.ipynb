# 06_Probability_Distributions

### What are Probability Distributions?

A Probability Distribution describes how the probabilities of different possible values or outcomes of a random variable are distributed.

Simply put, it tells us what outcomes are possible and how likely each outcome is.

Example

Suppose we toss a coin:

Head → Probability = 0.5
Tail → Probability = 0.5

The probability distribution tells us the probability associated with each possible outcome.

## 01. Bernoulli Distribution

**Definition**: The Bernoulli Distribution models a single trial with exactly two possible outcomes: success (1) or failure (0). It is the simplest discrete probability distribution.

**Formula**: P(X = 1) = p, P(X = 0) = 1 − p, where p = probability of success

**Where it is used**: Used to model any single yes/no, pass/fail, or true/false experiment — the building block for the Binomial Distribution.

**Real-World Example**: A single coin flip — heads (success) or tails (failure), with p = 0.5.

**AI/ML Example**: The output of a binary classifier (e.g., "fraud" vs "not fraud") for a single prediction is modeled as a Bernoulli random variable, and logistic regression predicts the parameter p.

```python
import numpy as np

p = 0.7
data = np.random.binomial(1, p, 1000)
print("Proportion of Successes:", np.mean(data))
```

**Interpretation**: About 70% of trials resulted in success (1), matching the set probability p=0.7 — confirming a single binary outcome trial.

## 02. Binomial Distribution

**Definition**: The Binomial Distribution models the number of successes in a fixed number of independent Bernoulli trials, each with the same probability of success.

**Formula**: P(X = k) = C(n, k) × p^k × (1−p)^(n−k)

where n = number of trials, k = number of successes, p = probability of success, C(n,k) = combinations

**Where it is used**: Used whenever an experiment is repeated a fixed number of times and each repetition is independent with the same success probability.

**Real-World Example**: The number of heads in 10 coin flips follows a Binomial Distribution.

**AI/ML Example**: Binomial distribution is used in A/B testing to model the number of conversions out of total visitors when comparing two versions of a webpage.

```python
import numpy as np
import matplotlib.pyplot as plt

n, p = 10, 0.5
data = np.random.binomial(n, p, 1000)

plt.hist(data, bins=range(0, n+2), rwidth=0.8, color='seagreen', edgecolor='black', align='left')
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()

print("Mean successes:", np.mean(data))
```

**Interpretation**: The histogram peaks around 5 successes (10 × 0.5), forming a roughly symmetric, bell-like shape since p=0.5 — the most likely outcome when repeating a 50/50 trial 10 times.

## 03. Poisson Distribution

**Definition**: The Poisson Distribution models the number of times an event occurs within a fixed interval of time or space, given a known average rate, when events happen independently.

**Formula**: P(X = k) = (λ^k × e^−λ) / k!

where λ = average rate of occurrence, k = number of occurrences

**Where it is used**: Used for modeling counts of rare or random events over a fixed interval — calls per hour, defects per batch, accidents per day.

**Real-World Example**: The number of customers arriving at a store in an hour, given an average of 10 customers/hour.

**AI/ML Example**: Poisson Regression is used to model count-based target variables, such as predicting the number of website visits or product purchases.

```python
import numpy as np

lam = 4
data = np.random.poisson(lam, 1000)
print("Mean events:", np.mean(data))
```

**Interpretation**: The average number of events observed is close to λ=4, confirming the simulated data matches the expected event rate typical of a Poisson process.

## 04. Uniform Distribution

**Definition**: The Uniform Distribution models a scenario where every value within a defined range is equally likely to occur. It can be discrete (dice roll) or continuous (random number between a and b).

**Formula**: f(x) = 1 / (b − a), for a ≤ x ≤ b (continuous case)

**Where it is used**: Used when there's no reason to believe any outcome is more likely than another within a range — random sampling, simulations, and initialization.

**Real-World Example**: The outcome of rolling a fair die — each face has an equal 1/6 chance.

**AI/ML Example**: Neural network weight initialization often draws values from a uniform distribution within a small range before training begins.

```python
import numpy as np

a, b = 1, 10
data = np.random.uniform(a, b, 1000)
print("Min:", np.min(data), "Max:", np.max(data))
print("Mean:", np.mean(data))
```

**Interpretation**: Values span the full range (1 to 10) with a mean close to the midpoint (~5.5), confirming every value in the range is equally likely to occur.

## 05. Normal Distribution

**Definition**: The Normal Distribution is a continuous, symmetric, bell-shaped distribution where values cluster around the mean, defined completely by mean (μ) and standard deviation (σ).

**Formula**: f(x) = (1 / (σ√(2π))) × e^(−(x−μ)² / 2σ²)

**Where it is used**: Used to model naturally occurring continuous data and forms the basis for statistical inference (confidence intervals, hypothesis testing, Z-scores).

**Real-World Example**: Heights of adult humans in a population approximately follow a normal distribution.

**AI/ML Example**: Many ML algorithms (Linear Regression, LDA, Gaussian Naive Bayes) assume normally distributed features or residuals for optimal performance.

```python
import numpy as np
import matplotlib.pyplot as plt

mean, sd = 70, 10
data = np.random.normal(mean, sd, 1000)

plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title(f"Normal Distribution (μ={mean}, σ={sd})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print("Mean:", np.mean(data), "SD:", np.std(data))
```

**Interpretation**: The histogram forms a symmetric bell curve centered at 70, with most values falling close to the mean and fewer values in the tails — the defining shape of a normal distribution.

## 06. Exponential Distribution

**Definition**: The Exponential Distribution models the time between events in a process where events occur continuously and independently at a constant average rate — it's the continuous counterpart to the Poisson distribution.

**Formula**: f(x) = λ × e^(−λx), for x ≥ 0

where λ = rate parameter (events per unit time)

**Where it is used**: Used for modeling "waiting times" — time until the next event, failure, or arrival.

**Real-World Example**: The time between customer arrivals at a store, or the time until a machine part fails.

**AI/ML Example**: Exponential distributions are used in survival analysis and reliability modeling to predict time-to-failure or time-to-churn for customers.

```python
import numpy as np

rate = 0.5
data = np.random.exponential(1/rate, 1000)
print("Mean wait time:", np.mean(data))
```

**Interpretation**: 
The average wait time is close to 1/λ = 2, confirming short waiting times are most common with a rapidly decreasing chance of longer waits — the "memoryless" decay pattern typical of exponential data.
