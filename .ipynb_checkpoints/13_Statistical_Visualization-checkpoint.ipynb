# 13_Statistical_Visualization

### What is Statistical Data Visualization?

Statistical data visualization is the practice of representing data graphically to reveal patterns, distributions, relationships, and outliers that are difficult to see in raw numbers. The right chart type depends on the data type (categorical, numerical, single or multiple variables) and the question being asked.

Example

A histogram quickly shows whether exam scores are normally distributed, while a scatter plot shows whether study hours and scores are related — the same data can need very different charts depending on the question.

## 01. Histogram

**Purpose**: A Histogram shows the distribution of a single numeric variable by dividing its range into bins and plotting the frequency of values falling into each bin.

**When to use**: Use when you want to understand the shape, spread, and skewness of a single continuous variable, such as checking for normality.

**Advantages**: Easy to interpret; clearly reveals distribution shape, central tendency, and spread; helps spot skewness and multiple peaks (modes).

**Limitations**: Sensitive to bin size/width choice, which can distort the perceived shape of the distribution; not suitable for comparing more than a couple of groups at once.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
scores = np.random.normal(70,10,200)
plt.hist(scores,bins=20,color='steelblue',edgecolor='black')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.title('Histogram of Exam Scores')
plt.show()
```

**Interpretation**: The histogram shows a roughly bell-shaped, symmetric distribution centered around 70, indicating exam scores are approximately normally distributed with no major skew.

## 02. Bar Chart

**Purpose**: A Bar Chart displays and compares values across discrete categories using rectangular bars, where the bar length or height represents the value.

**When to use**: Use when comparing a numeric measure (like sales, count, or average) across different categorical groups.

**Advantages**: Very intuitive and widely understood; easy to compare magnitudes across categories at a glance; works well for both few and moderately many categories.

**Limitations**: Becomes cluttered and hard to read with too many categories; does not show distribution or variability within each category, only a single summary value.

```python
import matplotlib.pyplot as plt
regions = ['North','South','East','West']
sales = [45000,38000,52000,41000]
plt.bar(regions,sales,color='coral')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.title('Sales by Region')
plt.show()
```

**Interpretation**: The East region has the highest sales among all four regions, while the South region lags behind, making regional performance differences immediately visible.

## 03. Line Chart

**Purpose**: A Line Chart connects data points in sequence with a line, primarily used to show trends and changes in a variable over a continuous interval, most commonly time.

**When to use**: Use when visualizing how a metric changes over time or another ordered continuous variable, such as tracking monthly revenue.

**Advantages**: Excellent for showing trends, patterns, and rate of change over time; can display multiple series together for easy trend comparison.

**Limitations**: Not suitable for categorical or unordered data; can become cluttered and misleading with too many overlapping lines or series.

```python
import numpy as np
import matplotlib.pyplot as plt
months = np.arange(1,13)
revenue = np.array([100,110,105,120,130,125,140,150,145,160,170,180])
plt.plot(months,revenue,marker='o',color='green')
plt.xlabel('Month')
plt.ylabel('Revenue (in thousands)')
plt.title('Monthly Revenue Trend')
plt.show()
```

**Interpretation**: The line chart shows a consistent upward trend in revenue over the year, with only minor dips in a couple of months, indicating overall business growth.

## 04. Scatter Plot

**Purpose**: A Scatter Plot displays individual data points on a two-dimensional graph based on two numeric variables, helping identify correlation, clusters, and outliers.

**When to use**: Use when examining the relationship between two continuous variables, such as checking if advertising spend relates to sales.

**Advantages**: Clearly reveals correlation direction and strength between two variables; easily highlights outliers and clusters within the data.

**Limitations**: Only shows relationships between two variables at a time; can become a dense, hard-to-read "cloud" with very large datasets (overplotting).

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
ad_spend = np.random.normal(5000,1000,100)
sales = ad_spend*3 + np.random.normal(0,2000,100)
plt.scatter(ad_spend,sales,color='purple',alpha=0.6)
plt.xlabel('Ad Spend')
plt.ylabel('Sales')
plt.title('Ad Spend vs Sales')
plt.show()
```

**Interpretation**: The scatter plot shows a clear positive linear relationship between ad spend and sales, indicating that higher advertising spend is generally associated with higher sales.

## 05. Pie Chart

**Purpose**: A Pie Chart represents the proportional share of categories within a whole, dividing a circle into slices sized according to each category's percentage contribution.

**When to use**: Use when showing how a small number of categories (ideally under 6-7) make up parts of a single total, such as market share breakdown.

**Advantages**: Intuitively communicates part-to-whole relationships to non-technical audiences; visually simple for a small number of categories.

**Limitations**: Becomes very hard to read and compare accurately with many categories or similar-sized slices; human eyes are poor at comparing angles/areas precisely, unlike bar lengths.

```python
import matplotlib.pyplot as plt
browsers = ['Chrome','Safari','Firefox','Edge','Other']
share = [65,18,8,6,3]
plt.pie(share,labels=browsers,autopct='%1.1f%%',startangle=90)
plt.title('Browser Market Share')
plt.show()
```

**Interpretation**: Chrome dominates the market with nearly two-thirds share, while Firefox, Edge, and Other browsers together make up a relatively small portion of the total.

## 06. Density Plot

**Purpose**: A Density Plot (KDE plot) shows a smoothed, continuous estimate of a variable's probability distribution, acting as a smoother alternative to a histogram.

**When to use**: Use when you want a smooth view of a distribution's shape, or when comparing the distributions of multiple groups on the same axis.

**Advantages**: Smooths out random noise from binning, making the underlying distribution shape clearer; multiple density curves overlay well for comparing groups.

**Limitations**: The smoothness depends on a bandwidth parameter, which can oversmooth or undersmooth the true distribution; can be less intuitive than a histogram for non-technical audiences.

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)
group_a = np.random.normal(60,8,200)
group_b = np.random.normal(75,8,200)
sns.kdeplot(group_a,label='Group A',fill=True)
sns.kdeplot(group_b,label='Group B',fill=True)
plt.xlabel('Score')
plt.title('Density Plot of Scores')
plt.legend()
plt.show()
```

**Interpretation**: The density plot shows Group B's scores are shifted noticeably higher than Group A's, with only a small overlap region between the two distributions.

## 07. Box Plot

**Purpose**: A Box Plot (Box-and-Whisker Plot) summarizes a numeric variable's distribution using its quartiles, displaying the median, interquartile range, and outliers, often across multiple groups.

**When to use**: Use when comparing the spread and central tendency of a numeric variable across several categories, or when checking for outliers.

**Advantages**: Compactly summarizes distribution and outliers; makes comparing spread and median across many groups easy in one chart.

**Limitations**: Does not show the full shape of the distribution (e.g., multiple peaks are hidden); can be less intuitive to beginners compared to a histogram.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
dept_a = np.random.normal(50000,8000,50)
dept_b = np.random.normal(60000,10000,50)
dept_c = np.random.normal(55000,5000,50)
plt.boxplot([dept_a,dept_b,dept_c],tick_labels=['Dept A','Dept B','Dept C'])
plt.ylabel('Salary')
plt.title('Salary Distribution by Department')
plt.show()
```

**Interpretation**: Department B has both the highest median salary and the widest spread, while Department C shows the most consistent (tightly clustered) salaries among the three departments.

## 08. Violin Plot

**Purpose**: A Violin Plot combines a box plot with a density plot, showing both summary statistics (median, quartiles) and the full shape of the distribution for one or more groups.

**When to use**: Use when comparing distributions across groups and the full shape of each distribution (including multiple peaks) matters, not just the summary statistics.

**Advantages**: Reveals more distributional detail than a box plot, such as bimodality; still allows easy comparison of spread and central tendency across groups.

**Limitations**: Can be harder for non-technical audiences to interpret than a simple box plot; density estimation can be misleading with very small sample sizes.

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)
dept_a = np.random.normal(50000,8000,50)
dept_b = np.random.normal(60000,10000,50)
dept_c = np.random.normal(55000,5000,50)
data = [dept_a,dept_b,dept_c]
plt.violinplot(data,showmedians=True)
plt.xticks([1,2,3],['Dept A','Dept B','Dept C'])
plt.ylabel('Salary')
plt.title('Violin Plot of Salary by Department')
plt.show()
```

**Interpretation**: The violin plot confirms the box plot's findings while also revealing that Department B's salary distribution is wider and slightly asymmetric, something not fully visible in the box plot alone.

## 09. Pair Plot

**Purpose**: A Pair Plot creates a grid of scatter plots (and histograms/density plots on the diagonal) for every pair of numeric variables in a dataset, giving a quick overview of all relationships at once.

**When to use**: Use during exploratory data analysis when you want to quickly scan relationships and distributions across several numeric features simultaneously.

**Advantages**: Efficiently summarizes pairwise relationships and individual distributions in a single figure; helps spot correlations, clusters, and outliers across many variable pairs at once.

**Limitations**: Becomes visually overwhelming and slow to render with many variables (grid grows quadratically); only captures pairwise, not higher-order, relationships.

```python
import numpy as np
import pandas as pd
import seaborn as sns
np.random.seed(42)
sqft = np.random.normal(1500,300,100)
rooms = sqft/300 + np.random.normal(0,0.5,100)
price = sqft*150 + np.random.normal(0,20000,100)
df = pd.DataFrame({'sqft':sqft,'rooms':rooms,'price':price})
sns.pairplot(df)
```

**Interpretation**: The pair plot shows a strong positive linear relationship between sqft and price, a similar relationship between sqft and rooms, and roughly normal distributions for each individual variable along the diagonal.

## 10. Heatmap

**Purpose**: A Heatmap uses color intensity to represent the magnitude of values in a matrix, commonly used to visualize correlation matrices or other tabular numeric data at a glance.

**When to use**: Use when visualizing a matrix of values, most commonly a correlation matrix, where color helps quickly identify strong positive, negative, or weak relationships.

**Advantages**: Makes patterns in large matrices instantly visible through color, much faster to scan than raw numbers; works well for correlation matrices and confusion matrices alike.

**Limitations**: Color perception can be subjective and misleading if the color scale is not chosen carefully; less precise than reading exact numeric values for detailed analysis.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)
sqft = np.random.normal(1500,300,100)
rooms = sqft/300 + np.random.normal(0,0.5,100)
age = np.random.normal(15,5,100)
price = sqft*150 + np.random.normal(0,20000,100)
df = pd.DataFrame({'sqft':sqft,'rooms':rooms,'age':age,'price':price})
sns.heatmap(df.corr(),annot=True,cmap='coolwarm',vmin=-1,vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

**Interpretation**: The heatmap clearly shows sqft and price are strongly positively correlated (dark red), while age has weak correlation with the other variables (near white), making key relationships immediately obvious through color.

## Overall Advantages and Limitations of Statistical Visualization

### Advantages
1. Makes patterns, trends, and outliers in data immediately visible, often faster than reading raw numbers or tables.
2. Different chart types are specialized for different data types and questions, giving flexibility to explore data from many angles.
3. Greatly improves communication of insights to both technical and non-technical audiences.
4. Helps validate statistical assumptions (like normality or linearity) before applying formal statistical tests or models.
5. Multi-variable visualizations like pair plots and heatmaps can summarize complex relationships in a single, digestible figure.

### Limitations
1. Choosing the wrong chart type for the data can misrepresent or obscure the actual pattern.
2. Visualizations can be manipulated (e.g., truncated axes, misleading scales) to create false impressions, intentionally or not.
3. Many charts (bar, pie) only show summary values and hide the underlying variability or distribution.
4. Visualizations become cluttered and less useful as the number of categories, variables, or data points grows very large.
5. Visual interpretation is somewhat subjective and should be supported with statistical tests for rigorous conclusions, not relied on alone.

## Interview Questions & Answers

**1. When would you choose a histogram over a density plot, or vice versa?**
A histogram is preferred when you want to see actual frequency counts and are fine with some sensitivity to bin size, while a density plot is preferred when you want a smoother view of the distribution shape or need to compare multiple overlapping distributions cleanly.

**2. What is the key difference between a box plot and a violin plot?**
A box plot only shows summary statistics (median, quartiles, outliers), while a violin plot combines those same summary statistics with a density curve showing the full shape of the distribution, revealing details like bimodality that a box plot would hide.

**3. Why is a pie chart often discouraged for datasets with many categories?**
Pie charts are discouraged with many categories because human perception is poor at comparing angles and areas precisely, making it hard to accurately judge and compare more than a few similarly-sized slices.

**4. What does a heatmap of a correlation matrix help you quickly identify?**
A correlation heatmap helps quickly identify which pairs of variables are strongly positively or negatively correlated through color intensity, making it easy to spot potential multicollinearity or key relationships without scanning raw numbers.

**5. When is a line chart not an appropriate choice for visualizing data?**
A line chart is not appropriate for unordered categorical data, since connecting points with a line implies a continuous trend or sequence that doesn't exist between unrelated categories, which can mislead the viewer.

**6. What is a major limitation of pair plots, and when would you avoid using one?**
A major limitation is that pair plots become visually overwhelming and computationally slow as the number of variables grows, since the grid size increases quadratically, so they should be avoided or reduced to key variables when working with datasets that have many numeric features.

**7. How can a scatter plot help before building a regression model?**
A scatter plot can visually reveal whether the relationship between two variables is linear, non-linear, or absent, and can highlight outliers, all of which help decide whether a linear regression model is appropriate before formally fitting one.
