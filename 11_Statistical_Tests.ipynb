### What are Statistical Tests?

Statistical tests are procedures used to determine whether observed differences or relationships in sample data are statistically significant or just due to random chance. Different tests are chosen based on data type, distribution, and the number of groups being compared.

Example

Comparing the average test scores of two different teaching methods uses a statistical test to check if the observed difference is real or coincidental.

## 01. T-Test

**Definition**: A T-Test compares the means of one or two groups to determine if they are statistically significantly different from each other or from a known value, using the t-distribution.

**Assumptions**: Data is approximately normally distributed; observations are independent; variances are equal between groups (for standard two-sample t-test); data is continuous.

**When to use**: Use when comparing means of one or two groups (or before/after pairs) and the sample size is relatively small with unknown population standard deviation.

**When not to use**: Do not use when data is heavily skewed or non-normal (especially with small samples), when comparing more than two groups, or when variances are very unequal without adjustment (use Welch's t-test instead).

**Business Example**: A company compares average customer satisfaction scores before and after a new support process to see if the change made a real difference.

```python
import numpy as np
from scipy import stats
before = np.array([70,72,68,74,71,69,73,70])
after = np.array([75,78,74,79,76,73,77,75])
t_stat,p_value = stats.ttest_rel(before,after)
print(t_stat)
print(p_value)
```

**Interpretation**: The p-value is well below 0.05, indicating a statistically significant improvement in satisfaction scores after the new support process was introduced.

## 02. Z-Test

**Definition**: A Z-Test compares a sample mean (or proportion) to a population value using the standard normal distribution, applicable when the population standard deviation is known and the sample size is large.

**Assumptions**: Data is approximately normally distributed; population standard deviation is known; sample size is large (typically n > 30); observations are independent.

**When to use**: Use when comparing a sample mean/proportion to a known population value, or comparing two large-sample means/proportions with known variances, such as in large-scale surveys or conversion rate testing.

**When not to use**: Do not use when the population standard deviation is unknown and sample size is small; a t-test is more appropriate in that case.

**Business Example**: An e-commerce company checks if the conversion rate of a new landing page (from a large sample of visitors) is significantly different from the known historical conversion rate of 5%.

```python
import numpy as np
from statsmodels.stats.weightstats import ztest
np.random.seed(1)
sample = np.random.normal(0.06,0.02,500)
z_stat,p_value = ztest(sample,value=0.05)
print(z_stat)
print(p_value)
```

**Interpretation**: The very small p-value indicates the new landing page's conversion rate is statistically significantly different from the historical 5% baseline.

## 03. Chi-Square Test

**Definition**: A Chi-Square Test examines whether there is a significant association between two categorical variables, or whether observed frequencies differ significantly from expected frequencies.

**Assumptions**: Data consists of categorical frequencies/counts (not raw continuous values); observations are independent; expected frequency in each cell is generally at least 5.

**When to use**: Use when testing the relationship between two categorical variables (test of independence) or comparing observed vs expected category distributions (goodness of fit).

**When not to use**: Do not use with continuous data without binning it, or when expected cell counts are too small, which makes the test unreliable.

**Business Example**: A retailer checks if there is an association between customer gender and preferred payment method (cash, card, wallet) using survey data.

```python
import numpy as np
from scipy import stats
table = np.array([[120,90,40],[80,110,60]])
chi2,p_value,dof,expected = stats.chi2_contingency(table)
print(chi2)
print(p_value)
```

**Interpretation**: The low p-value suggests a statistically significant association between gender and preferred payment method, meaning the two variables are not independent.

## 04. ANOVA

**Definition**: ANOVA (Analysis of Variance) tests whether there is a statistically significant difference between the means of three or more independent groups, using the F-distribution.

**Assumptions**: Data is approximately normally distributed within each group; groups have roughly equal variances (homogeneity of variance); observations are independent.

**When to use**: Use when comparing means across three or more groups simultaneously, such as testing multiple marketing campaigns or treatment groups at once.

**When not to use**: Do not use when comparing only two groups (a t-test is sufficient), or when data is non-normal or variances are highly unequal (use Kruskal-Wallis instead).

**Business Example**: A company compares average sales performance across three different regional teams to see if any region significantly outperforms the others.

```python
import numpy as np
from scipy import stats
region_a = np.array([200,210,195,205,198])
region_b = np.array([220,225,230,215,218])
region_c = np.array([180,175,190,185,178])
f_stat,p_value = stats.f_oneway(region_a,region_b,region_c)
print(f_stat)
print(p_value)
```

**Interpretation**: The very small p-value indicates a statistically significant difference in average sales performance among at least one pair of the three regional teams.

## 05. Mann-Whitney U Test

**Definition**: The Mann-Whitney U Test is a non-parametric test that compares whether two independent groups come from the same distribution, based on ranking the combined data rather than assuming normality.

**Assumptions**: Observations are independent; data is at least ordinal; does not require normal distribution, making it a robust alternative to the t-test.

**When to use**: Use when comparing two independent groups but the data is not normally distributed, contains outliers, or is ordinal rather than continuous.

**When not to use**: Do not use when data is normally distributed and a t-test's higher statistical power can be leveraged instead, or when comparing more than two groups.

**Business Example**: A company compares customer satisfaction ratings (on a 1-10 ordinal scale) between two service centers where the ratings are skewed rather than normally distributed.

```python
import numpy as np
from scipy import stats
center_a = np.array([8,9,7,9,8,10,6,9])
center_b = np.array([6,5,7,6,5,8,4,6])
u_stat,p_value = stats.mannwhitneyu(center_a,center_b)
print(u_stat)
print(p_value)
```

**Interpretation**: The small p-value indicates a statistically significant difference in satisfaction ratings between the two service centers, without assuming the ratings are normally distributed.

## 06. Shapiro-Wilk Test

**Definition**: The Shapiro-Wilk Test checks whether a sample of data comes from a normally distributed population, and is one of the most powerful normality tests for small to moderate sample sizes.

**Assumptions**: Observations are independent and continuous; primarily reliable for sample sizes roughly between 3 and 5000.

**When to use**: Use before applying tests that assume normality (like t-test, ANOVA, or Z-test) to verify whether that assumption holds for the data.

**When not to use**: Do not rely on it alone for very large samples, where even trivial deviations from normality can produce a significant result; visual checks (Q-Q plots) should also be used.

**Business Example**: Before running a t-test to compare delivery times of two logistics providers, an analyst uses the Shapiro-Wilk test to confirm the delivery time data is approximately normal.

```python
import numpy as np
from scipy import stats
np.random.seed(1)
delivery_times = np.random.normal(48,5,30)
stat,p_value = stats.shapiro(delivery_times)
print(stat)
print(p_value)
```

**Interpretation**: The p-value is well above 0.05, so there is no significant evidence against normality, meaning the delivery time data can reasonably be treated as normally distributed for further tests.

## Overall Advantages and Limitations of Statistical Tests

### Advantages
1. Provide an objective, quantifiable way to validate business and research decisions using data.
2. Cover a wide range of scenarios: parametric (t-test, z-test, ANOVA) and non-parametric (Mann-Whitney) options exist for different data types.
3. Help detect real relationships (Chi-Square) or differences (ANOVA, t-test) versus random noise.
4. Normality tests like Shapiro-Wilk help validate assumptions before choosing the right test, improving reliability of conclusions.
5. Widely supported in statistical software and libraries, making them easy to apply and reproduce.

### Limitations
1. Choosing the wrong test for the data type or distribution can lead to invalid conclusions.
2. Most parametric tests are sensitive to violations of normality and equal variance assumptions.
3. Statistical significance does not always mean practical or business significance.
4. Non-parametric tests like Mann-Whitney generally have less statistical power than their parametric counterparts when normality actually holds.
5. Shapiro-Wilk and similar normality tests become overly sensitive with very large sample sizes, flagging trivial deviations as significant.

## Interview Questions & Answers

**1. What is the key difference between a t-test and a z-test?**
A t-test is used when the population standard deviation is unknown and sample sizes are typically small, using the t-distribution, while a z-test is used when the population standard deviation is known and sample sizes are large, using the standard normal distribution.

**2. When would you choose ANOVA over multiple t-tests?**
ANOVA is used when comparing means across three or more groups at once, because running multiple pairwise t-tests inflates the overall risk of a Type I error, whereas ANOVA controls this risk with a single test.

**3. What is the purpose of the Chi-Square test, and what type of data does it require?**
The Chi-Square test checks for association between categorical variables or compares observed versus expected frequencies, and it requires categorical count data rather than continuous values.

**4. Why would you use the Mann-Whitney U test instead of a t-test?**
The Mann-Whitney U test is used instead of a t-test when the data is not normally distributed, contains outliers, or is only ordinal, since it compares distributions using ranks rather than assuming normality.

**5. What does the Shapiro-Wilk test tell you, and why is it important before running other tests?**
The Shapiro-Wilk test checks whether a sample comes from a normally distributed population, and it's important because many common tests like the t-test and ANOVA assume normality, so verifying this first ensures the chosen test is appropriate.

**6. What assumption do ANOVA and the t-test share regarding variance?**
Both tests generally assume homogeneity of variance, meaning the groups being compared should have roughly equal variances; violating this assumption can require corrections like Welch's t-test or Welch's ANOVA.

**7. If a Shapiro-Wilk test shows data is not normally distributed, what should you do next?**
If data fails the normality assumption, you should consider using a non-parametric alternative, such as the Mann-Whitney U test instead of a t-test, or transforming the data (e.g., log transformation) to make it closer to normal.
